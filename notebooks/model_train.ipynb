```python
from ultralytics import YOLO
import wandb
import os

# Initialize your Weights & Biases environment
wandb.login(key=os.getenv("WANDB_API_KEY"))
```

```python
model = YOLO("../models/yolo26s.pt")
```

```python
results = model.train(data="../dataset/data.yaml", 
                      epochs=1, 
                      # imgsz=640,
                      batch=4,
                      device="mps",
                      project="football_detection",
                      name="football_detection_v0",

                      hsv_h=0.015,
                      hsv_s=0.7,
                      hsv_v=0.4,
                      degrees=0.0,
                      translate=0.1,
                      scale=0.5,
                      shear=0.0,
                      perspective=0.0,
                      flipud=0.0,
                      fliplr=0.5,
                      bgr=0.0,
                      mosaic=1.0,
                      mixup=0.0,
                      cutmix=0.0,
                    #   copy_paste=0.0,
                    #   copy_paste_mode=0,
                      auto_augment='randomaugment',
                      erasing=0.4,
                      augmentations='detect'
                    )
```
